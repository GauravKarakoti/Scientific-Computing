```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def function(x):
  return np.sin(x)    # Example function
  # return np.exp(2*x)
```

```python
def forward_difference(f, x, h):
  return (f(x+h) - f(x)) / h
```

```python
def backward_difference(f, x, h):
  return (f(x) - f(x-h)) / h
```

```python
def central_difference(f, x, h):
  return (f(x+h) - f(x-h)) / (2*h)
```

```python
def second_derivative(f, x, h):
  return (f(x+h) - 2*f(x) + f(x-h)) / h**2
```

```python
# Define x values for differentiation
x_vals = np.linspace(0, 2*np.pi, 100)
h=0.01
```

```python
# Compute derivatives
forward_diff = forward_difference(function, x_vals, h)
backward_diff = backward_difference(function, x_vals, h)
central_diff = central_difference(function, x_vals, h)
second_deriv = second_derivative(function, x_vals, h)
```

```python
# Exact derivative for comparison
exact_derivative = np.cos(x_vals)
# exact_derivative = 2*np.exp(2*x_vals)
```

```python
# Plot results
plt.figure(figsize=(10, 6))
plt.plot(x_vals, exact_derivative, label='Exact Derivative (cos(x))', linestyle='dashed')
plt.plot(x_vals, forward_diff, label='Forward Difference', alpha=0.7)
plt.plot(x_vals, backward_diff, label='Backward Difference', alpha=0.7)
plt.plot(x_vals, central_diff, label='Central Difference', alpha=0.7)
plt.legend()
plt.title("Numerical Differentiation of sin(x)")
# plt.title("Numerical Differentiation of 2exp(2x)")
plt.xlabel('x')
plt.ylabel('Derivative')
plt.grid()
plt.show()
```
