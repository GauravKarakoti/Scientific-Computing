```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def trapezoidal_rule(a, b, n, discrete_data):
  x_points = np.linspace(a, b, len(discrete_data))
  dx = (b-a) / (len(discrete_data)-1)
  y_values = np.array(discrete_data)
  sum_value = (dx/2) * (2*np.sum(y_values[1: -1])+y_values[0]+y_values[-1])
  return sum_value, x_points, dx, y_values
```

```python
def plot_trapezoidal(a, b, n, discrete_data):
  sum_value, x_points, dx, y_values = trapezoidal_rule(a, b, n, discrete_data)
  plt.figure(figsize=(8, 5))
  plt.scatter(x_points, y_values, color='red', label='Discrete Data')
  plt.plot(x_points, y_values, 'b-', alpha=0.6)
  for i in range(len(x_points)-1):
    plt.fill_between([x_points[i], x_points[i+1]], [y_values[i], y_values[i+1]], alpha=0.4, color='blue', edgecolor='black')
  plt.title(f'Trapezoidal Rule, n={len(discrete_data)-1}\nApproximate Integral: {sum_value:.5f}')
  plt.xlabel('x')
  plt.ylabel('f(x)')
  plt.legend()
  plt.show()
```

```python
# Parameters
# a, b = 0, np.pi    # Interval
a, b = 0, 1.2
n = 15    # Number of points
# -------------For a given function-------------
discrete_x = np.linspace(a, b, n)
# discrete_y = np.sin(discrete_x) + np.random.normal(0, 0.05, len(discrete_x))    # Noise sine wave
discrete_y = np.exp(-(discrete_x**2))
discrete_data = list(discrete_y)
```

```python
plot_trapezoidal(a, b, n, discrete_data)
```

```python
# -------------Example with given data points-------------
# given_x = np.array([1, 2, 3, 4, 5])
# given_y = np.array([1, 4, 7, 14, 20])
given_x = np.array([0, 0.2, 0.4, 0.6, 0.8, 1, 1.2])
given_y = np.array([1, 0.9608, 0.8521, 0.6977, 0.5273, 0.3679, 0.2369])
plot_trapezoidal(given_x[0], given_x[-1], len(given_x), given_y)
```
