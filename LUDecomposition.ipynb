```python
import numpy as np
```

```python
def lu_gauss_elimination(A):
  n = len(A)
  L = np.eye(n)          # Initialize L as identity matrix
  U = A.astype(float)    # Copy A to U
  for i in range(n):
    for j in range(i+1, n):
      factor = U[j, i] / U[i, i]
      L[j, i] = factor
      U[j] = U[j] - factor * U[i]
  return L, U
```

```python
def lu_crout(A):
  n = len(A)
  L = np.zeros((n, n))
  U = np.eye(n)
  for i in range(n):
    for j in range(i, n):
      L[j, i] = A[j, i] - sum(L[j, k] * U[k, i] for k in range(i))
    for j in range(i+1, n):
      U[i, j] = (A[i, j] - sum(L[i, k] * U[k, j] for k in range(i))) / L[i, i]
  return L, U
```

```python
# Example Usage
# A = np.array([[4, -2, 4], [-2, 10, -2], [4, -2, 6]], dtype=float)
A = np.array([[3, 2, 1, -4], [1, -5, 2, 1], [5, 1, -3, 2], [2, 3, 1, 5]], dtype=float)
```

```python
L_gauss, U_gauss = lu_gauss_elimination(A)
L_crout, U_crout = lu_crout(A)
```

```python
print("LU Decomposition using Gauss Elimination")
print("L:")
print(L_gauss)
print("U:")
print(U_gauss)
```

```python
print("\nLU Decomposition using Crout's Elimination")
print("L:")
print(L_crout)
print("U:")
print(U_crout)
```
