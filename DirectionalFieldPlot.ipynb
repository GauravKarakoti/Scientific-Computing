```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def direction_field(f, x_range, y_range, density=20, title='Direction Field'):
  x = np.linspace(x_range[0], x_range[1], density)
  y = np.linspace(y_range[0], y_range[1], density)
  X, Y = np.meshgrid(x, y)
  # Calculate slopes at each grid point
  U = 1
  V = f(X, Y)
  # Normalize the vectors for better visualization
  N = np.sqrt(U**2 + V**2)
  U, V = U/N, V/N
  plt.figure(figsize=(8, 6))
  plt.quiver(X, Y, U, V, angles='xy')
  plt.xlabel('x')
  plt.ylabel('y')
  plt.title(title)
  plt.grid(True)
  plt.show()
```

```python
def differential_eq1(x, y):
  return x**2 + y**2 - 1
```

```python
def differential_eq2(x, y):
  return 1 + x * y**2
```

```python
# Define the range for x and y
x_range = (-2, 2)
y_range = (-2, 2)
```

```python
# Plotting the first direction field
direction_field(differential_eq1, x_range, y_range, title='Direction Field for dy/dx = x^2+y^2-2')
```

```python
# Plotting the second direction field
direction_field(differential_eq2, x_range, y_range, title='Direction Field for dy/dx = 1+x*y^2')
```
